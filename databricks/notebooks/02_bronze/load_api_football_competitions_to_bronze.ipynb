```python
from pyspark.sql import functions as F
import json
```

```python
file_path = "/Volumes/football_data_platform/raw/source_files/api_football/competition_data/raw_competitions.json"


bronze_df = (
    spark.read
    .option("multiLine", "true")
    .json(file_path)
    .withColumn("_ingested_at", F.current_timestamp())
    .withColumn("_source_file", F.lit(file_path))
    .withColumn("_source_system", F.lit("API-Football"))
)
(
bronze_df.write
.format("delta")
.mode("append")
.option("mergeSchema", "true")
.saveAsTable("football_data_platform.bronze.api_competitions_data")
)
```
