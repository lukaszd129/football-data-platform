```python
import pandas as pd
import json
```

```python
with open("/Volumes/football_data_platform/raw/source_files/api_football/player_statistics/test_raw_player_season.json") as f:
  data = json.load(f)
print(json.dumps(data, indent=4))
```

```python
df = pd.json_normalize(
    data["response"],
    record_path=["statistics"],
    meta=[
        ["player", "id"],
        ["player", "name"],
        ["player", "firstname"],
        ["player", "lastname"],
        ["player", "age"],
        ["player", "birth", "date"],
        ["player", "birth", "place"],
        ["player", "birth", "country"],
        ["player", "nationality"],
        ["player", "height"],
        ["player", "weight"],
        ["player", "injured"]
    ]
)

pd.set_option("display.max_columns", None)

print(df)
```
