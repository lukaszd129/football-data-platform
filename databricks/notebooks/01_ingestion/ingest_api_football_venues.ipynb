```python
import requests
import pandas as pd
import os
import json
```

```python
countries = ['England','Italy','Germany','Spain','France']
venues_url = "https://v3.football.api-sports.io/venues"
api_key = '1b8444b9cb616e2c06743ebd3a22a2ff'
headers = {'x-apisports-key': api_key}
```

```python
session=requests.Session()
```

```python
#VENUES DATA
all_venues = []
for country in countries:
    response = session.get(
        url = venues_url,
        params = {"country":country},
        headers = headers,
        timeout = 30
    )

    response.raise_for_status()

    data = response.json()
    all_venues.extend(data["response"])
filename = (
    f"/Volumes/football_data_platform/raw/source_files/"
    f"api_football/venue_data/raw_venues.json"
)
os.makedirs(os.path.dirname(filename), exist_ok=True)

with open(filename, "w", encoding="utf-8") as file:
    json.dump(all_venues, file, indent=2)
```
