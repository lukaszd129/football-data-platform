```python
import requests
import pandas as pd
import os
import json
import time
```

```python
players_url = "https://v3.football.api-sports.io/players"
leagues = {
    '61':'league_one',
    '71':'serie_a',
    '39':'premier_league',
    '78':'bundesliga',
    '140':'la_liga'    
}
seasons = ['2021','2022','2023','2024','2025']

api_key = '1b8444b9cb616e2c06743ebd3a22a2ff'
headers = {'x-apisports-key': api_key}
```

```python
session=requests.Session()
```

```python
#PLAYER SEASON STATS
all_stats = []
for league_id, league_name in leagues.items():
    for season in seasons:

        print(f"Fetching {league_name} - {season}")

        while True:
            response = session.get(
                url=players_url,
                params={
                    "league": league_id,
                    "season": season
                },
                headers=headers,
                timeout=30
            )

            if response.status_code == 429:
                print("Rate limit reached. Waiting 60 seconds...")
                time.sleep(60)
                continue

            response.raise_for_status()
            break

        data = response.json()

        print(f"Retrieved {len(data['response'])} players")

        all_stats.extend(data["response"])

        # Small delay between successful requests
        time.sleep(1)


filename = (
    "/Volumes/football_data_platform/raw/source_files/"
    "api_football/player_statistics/raw_player_season_stats.json"
)

os.makedirs(os.path.dirname(filename), exist_ok=True)

with open(filename, "w", encoding="utf-8") as file:
    json.dump(all_stats, file, indent=2)
```
