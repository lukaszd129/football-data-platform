```python
import pandas as pd
import json
```

```python
with open("/Volumes/football_data_platform/raw/source_files/api_football/coach_data/test_raw_coach.json") as f:
  data = json.load(f)
print(json.dumps(data, indent=4))
```

```python
df = pd.json_normalize(
    data["response"],
    record_path=["career"],
    meta=[
        "id",
        "name",
        "firstname",
        "lastname",
        "age",
        "nationality",
        "height",
        "weight",
        "photo",
        "career"
    ]
    )
#df = pd.json_normalize(df["career"])
df = df[["start", "end", "team.id", "team.name", "id", "name", "firstname", "lastname", "age", "nationality", "height", "weight"]]


print(df.to_string())
```
